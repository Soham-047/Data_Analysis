# 📊 Learning Matplotlib
This notebook introduces **Matplotlib**, a powerful Python library for data visualization.

We'll cover:
- Basic plots (line, bar, scatter, histogram, pie)
- Customization (colors, labels, legends)
- Subplots & multiple figures
- Advanced visualizations
- Using Matplotlib with NumPy & Pandas

```python

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Inline plotting for Jupyter
%matplotlib inline
```

## 1. Line Plot

```python

x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.plot(x, y, label="sin(x)", color="blue", linestyle="--")
plt.title("Line Plot Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend()
plt.show()
```

## 2. Bar Chart

```python

categories = ['A', 'B', 'C', 'D']
values = [4, 7, 2, 9]

plt.bar(categories, values, color="orange")
plt.title("Bar Chart Example")
plt.show()
```

## 3. Scatter Plot

```python

x = np.random.rand(50)
y = np.random.rand(50)

plt.scatter(x, y, color="green", marker="o")
plt.title("Scatter Plot Example")
plt.show()
```

## 4. Histogram

```python

data = np.random.randn(1000)

plt.hist(data, bins=30, color="purple", alpha=0.7)
plt.title("Histogram Example")
plt.show()
```

## 5. Pie Chart

```python

sizes = [15, 30, 45, 10]
labels = ["A", "B", "C", "D"]
colors = ["red", "blue", "green", "yellow"]

plt.pie(sizes, labels=labels, colors=colors, autopct="%1.1f%%")
plt.title("Pie Chart Example")
plt.show()
```

## 6. Subplots

```python

x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)

fig, axs = plt.subplots(2, 1, figsize=(6, 8))

axs[0].plot(x, y1, "r-")
axs[0].set_title("sin(x)")

axs[1].plot(x, y2, "b-")
axs[1].set_title("cos(x)")

plt.tight_layout()
plt.show()
```

## 7. Advanced Plots

```python

# Stacked Bar
categories = ['X1', 'X2', 'X3']
values1 = [3, 5, 7]
values2 = [2, 4, 6]

plt.bar(categories, values1, label="Group 1")
plt.bar(categories, values2, bottom=values1, label="Group 2")
plt.title("Stacked Bar Chart")
plt.legend()
plt.show()

# Filled Area Plot
x = np.linspace(0, 10, 200)
y1 = np.sin(x)
y2 = np.sin(x) + 1

plt.fill_between(x, y1, y2, color="skyblue", alpha=0.4)
plt.title("Filled Area Plot")
plt.show()
```

## 8. Using Matplotlib with Pandas

```python

# Create a sample DataFrame
data = {
    "Year": [2018, 2019, 2020, 2021, 2022],
    "Sales": [200, 300, 400, 350, 500]
}
df = pd.DataFrame(data)

df.plot(x="Year", y="Sales", kind="line", marker="o", title="Sales Growth", legend=False)
plt.ylabel("Sales")
plt.show()
```
