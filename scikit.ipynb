```python
import sklearn
from sklearn.datasets import load_iris
from matplotlib import pyplot as plt
```

```python
X, y = load_iris(return_X_y=True)
```

```python
X, y
```

```python
from sklearn.neighbors import KNeighborsRegressor
from sklearn.linear_model import LinearRegression
```

```python
modknn = KNeighborsRegressor()
modlin = LinearRegression()
```

```python
modknn.fit(X, y)
```

```python
modknn.predict(X)
```

```python
modlin.fit(X, y)
```

```python
pred = modlin.predict(X)
```

```python
plt.scatter(pred, y)
```

```python
pred2 = modknn.predict(X)
```

```python
plt.scatter(pred2, y)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
pipe = Pipeline([
    ("scale", StandardScaler()),
    ("model", KNeighborsRegressor())
])
```

```python
pipe
```

```python
pipe.fit(X, y)
```

```python
pred = pipe.predict(X)
```

```python
plt.scatter(pred, y)
```

```python
pipe.get_params()
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
mod = GridSearchCV(
    estimator = pipe,
    param_grid={
        "model__n_neighbors": [1, 3, 5, 7, 9, 11, 13, 15]
    },
    cv = 3
)
```

```python
mod.fit(X, y)
```

```python
mod.cv_results_
```

```python
import pandas as pd
```

```python
pd.DataFrame(mod.cv_results_)
```
